# Validation

See how the data fares against the dataset's rules. Reads are safe; `validate_and_wait` starts a new validation run (it writes). Ids: see the [quickstart](quickstart.ipynb).

```python
# Only needed if reportnet_api is not installed yet:
%pip install -q "reportnet-api[pandas]"
```

```python
from reportnet_api.easy import connect_interactive, check_access

client = connect_interactive()  # hidden prompt, nothing is saved in the notebook
check_access(client)
```

```python
DATAFLOW_ID = None  # <- your dataflow id
DATASET_ID = None   # <- a dataset id (NOT the dataflow id)
```

## Summary and error list
On a data-lake dataflow some plain reads answer empty; if the table here is empty but the web app shows failures, check `easy.is_data_lake(client, DATAFLOW_ID)` (see the glossary).

```python
from reportnet_api.easy import validation_summary, validation_errors

validation_summary(client, DATASET_ID, DATAFLOW_ID)
```

```python
validation_errors(client, DATASET_ID, DATAFLOW_ID).head(20)
```

## Save the failures as a CSV

```python
from reportnet_api.easy import export_validations

export_validations(client, DATASET_ID, "validation_errors.csv")
```
