# Compare countries and save to Excel

Read the same table from several countries' datasets, stack them, summarise and save an Excel
file. Needs an **ApiKey** for the dataflow (reads of data tables are not allowed for every
login token, see the [credentials guide](../docs/credentials.md)). Reads only; nothing is changed.

```python
# Only needed if reportnet_api is not installed yet:
%pip install -q "reportnet-api[pandas]"
```

```python
%pip install -q openpyxl  # only needed for the Excel file at the end
```

```python
from reportnet_api.easy import connect_interactive, find_dataflow, read_many

client = connect_interactive()
```

## 1. Pick your dataflow
Use the id or part of its name.

```python
flow = find_dataflow(client, "PUT PART OF THE NAME OR THE ID HERE")
flow.datasets
```

## 2. Choose the datasets and the table
Country codes work as names: `flow.dataset("AT")`. `tables` lists the table names of one dataset.

```python
flow.dataset("AT").tables  # replace AT by one of the countries above
```

```python
countries = ["AT", "BE"]  # <- the countries you want to compare
TABLE = "PUT THE TABLE NAME HERE"

all_rows = read_many([flow.dataset(c) for c in countries], TABLE, dtypes="schema")
all_rows.head()
```

## 3. Summarise
Replace `COLUMN` with a column of your table.

```python
all_rows.groupby("country").size()  # rows per country
```

```python
# all_rows.groupby("country")["COLUMN"].describe()
```

## 4. Save to Excel

```python
all_rows.to_excel("comparison.xlsx", index=False)
"saved comparison.xlsx"
```

```python
client.close()
```
