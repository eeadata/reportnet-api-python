# Try it without a credential

This notebook uses **built-in fictional data**: no key, no network, nothing can be changed.
It shows the same steps as the real [quickstart](quickstart.ipynb), so you can learn the
functions first. Every name and number below is invented (dataflow 1, datasets 101 to 103).

```python
# Only needed if reportnet_api is not installed yet:
%pip install -q "reportnet-api[pandas] @ git+https://github.com/eeadata/reportnet-api-python"
```

## 1. Connect to the demo

```python
from reportnet_api import easy

client = easy.demo_client()
```

## 2. Find dataflows and datasets
With a real credential these are *your* dataflows. Note that a dataflow and its datasets have **different ids**.

```python
easy.list_dataflows(client)[["id", "name", "status"]]
```

```python
easy.list_datasets(client, 1)
```

## 3. Tables, and one table as a DataFrame
`dtypes="schema"` turns numbers and dates into real numbers and dates, using the dataset's schema.

```python
easy.list_tables(client, 102)
```

```python
austria = easy.read_table(client, 102, 1, dtypes="schema")
austria
```

```python
austria.dtypes
```

## 4. Several countries at once
`find_dataflow` gives a handle that remembers the ids; `flow.dataset("AT")` picks a dataset by country code.

```python
flow = easy.find_dataflow(client, "demo")
flow
```

```python
both = easy.read_many([flow.dataset("AT"), flow.dataset("BE")], "Stations", dtypes="schema")
both
```

```python
both.groupby("country")["Station"].count()
```
