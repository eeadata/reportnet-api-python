# Export and upload

Save data to files and load a file into a dataset. See the [quickstart](quickstart.ipynb) for finding ids. **Uploads write to the dataset: use a test or scratch dataset.**

```python
# Only needed if reportnet_api is not installed yet:
%pip install -q "reportnet-api[pandas]"
```

```python
from reportnet_api.easy import connect_interactive, check_access

client = connect_interactive()  # hidden prompt, nothing is saved in the notebook
check_access(client)
```

```python
DATAFLOW_ID = None  # <- your dataflow id
DATASET_ID = None   # <- a dataset id (NOT the dataflow id)
```

## Export one table
The file format can be `csv`, `xlsx`, ... The call waits until the file is ready.

```python
from reportnet_api.easy import export_table

export_table(client, DATASET_ID, "MyTable", "my_table.csv")
```

## Export the whole dataset
A ZIP of CSV files (or `file_format="parquet"`). It runs as a server job, so it can take a minute; `progress=True` shows the status. An ApiKey is needed for these exports (see the glossary and reference).

```python
from reportnet_api.easy import export_to_file

export_to_file(client, DATASET_ID, DATAFLOW_ID, "dataset.zip", progress=True)
```

## Upload a file
The file name must match a table name (`MyTable.csv`), otherwise the import job is cancelled. `replace=True` deletes existing data first: when live-tested it also emptied the dataset's *other* tables, so use it only on a test dataset, and it needs `confirm=True`. If the delimiter is not a comma, pass `delimiter="|"`.

```python
from reportnet_api.easy import import_file

import_file(client, DATASET_ID, DATAFLOW_ID, "MyTable.csv", progress=True)
```

```python
client.close()
```
